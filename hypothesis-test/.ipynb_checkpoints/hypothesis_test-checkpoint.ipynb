# Footedness & Finishing Efficiency in Football (2014–2020)

This notebook tests the belief that left-footed players finish better than right-footed players using real shot-level data from Europe’s top leagues (2014–2020).  
We compute player finishing efficiency, run a permutation test, bootstrap uncertainty, and also validate results by breaking performance down by strong foot vs weak foot.

## 2. Data Filtering, Feature Engineering, and Aggregation

In this section, I prepare the dataset for analysis by filtering shot events, creating efficiency metrics, and aggregating performance at the player level.

### **2.1 Shot-Level Filtering**
To ensure a fair comparison of finishing ability, I restrict the dataset to comparable, open-play footed shots:

- Keep only RightFoot and LeftFoot shots  
- Keep only OpenPlay situations  
- Remove shots with outcomes OwnGoal or ShotOnPost

### **2.2 Creating Performance Variables**
I create two key variables:

- **goal** → 1 if the shot was scored, else 0  
- **efficiency** → `goal - xG` (how much the player over- or under-performed expectations)

### **2.3 Strong-Foot Indicator**
To analyze strong vs weak foot performance, I create:

- `is_strong_foot = 1` when  
  - Right-footed player takes a RightFoot shot  
  - Left-footed player takes a LeftFoot shot  
- otherwise `0`

### **2.4 Foot-by-Foot Aggregation (RightFoot vs LeftFoot)**
For each player and each foot, I calculate:

- number of shots  
- number of goals  
- total xG  
- total efficiency  
- average efficiency  

These are pivoted into a wide format to produce columns such as:  
`shots_RightFoot`, `goals_LeftFoot`, `avg_eff_RightFoot`, etc.

### **2.5 Strong vs Weak Foot Aggregation**
Similarly, I aggregate performance by strong vs weak foot:

- `shots_strong`, `goals_strong`, `xg_strong`, `avg_eff_strong`  
- `shots_weak`, `goals_weak`, `xg_weak`, `avg_eff_weak`

Only players with ≥ 10 total strong+weak shots are kept.

### **2.6 Player-Level Finishing Efficiency**
Separately, I compute overall finishing efficiency per player:

- total shots  
- total goals  
- total xG  
- total efficiency  
- avg_efficiency (mean efficiency across all footed open-play shots)

Only players with **≥ 10 total shots** are included in the final analysis dataset.

This filtered and aggregated dataset forms the basis for the permutation test, bootstrap CI, and all subsequent hypothesis evaluation.


```python
df = df[df['shotType'].isin(['RightFoot', 'LeftFoot'])]
df = df[df['situation'] == 'OpenPlay']
df = df[~df['shotResult'].isin(['OwnGoal', 'ShotOnPost'])]
```

```python
df['goal'] = df['shotResult'].apply(lambda x: 1 if x == 'Goal' else 0)
df['efficiency'] = df['goal'] - df['xGoal']
```

```python
df['is_strong_foot'] = (
    ((df['preferred_foot'] == 'Right') & (df['shotType'] == 'RightFoot')) |
    ((df['preferred_foot'] == 'Left')  & (df['shotType'] == 'LeftFoot'))
).astype(int)
```

```python
foot_eff = df.groupby(['name', 'shotType']).agg(
    shots=('goal', 'count'),
    goals=('goal', 'sum'),
    xg=('xGoal', 'sum'),
    efficiency=('efficiency', 'sum'),
    avg_eff=('efficiency', 'mean')
).reset_index()
```

```python
foot_eff_pivot = foot_eff.pivot_table(
    index='name',
    columns='shotType',
    values=['shots', 'goals', 'xg', 'efficiency', 'avg_eff'],
    fill_value=0
)


foot_eff_pivot.columns = [
    f"{metric}_{foot}" for metric, foot in foot_eff_pivot.columns
]

foot_eff_pivot = foot_eff_pivot.reset_index()
foot_eff_pivot = foot_eff_pivot.merge(player_info, on='name', how='left')
```

```python
strong_weak = df.groupby(['name', 'is_strong_foot']).agg(
    shots=('goal', 'count'),
    goals=('goal', 'sum'),
    xg=('xGoal', 'sum'),
    efficiency=('efficiency', 'sum'),
    avg_eff=('efficiency', 'mean')
).reset_index()
```

```python
strong_weak_pivot = strong_weak.pivot_table(
    index='name',
    columns='is_strong_foot',
    values=['shots', 'goals', 'xg', 'efficiency', 'avg_eff'],
    fill_value=0
)

strong_weak_pivot.columns = [
    f"{metric}_{'strong' if foot==1 else 'weak'}"
    for metric, foot in strong_weak_pivot.columns
]

strong_weak_pivot = strong_weak_pivot.reset_index()
strong_weak_pivot = strong_weak_pivot.merge(player_info, on='name', how='left')
```

```python
strong_weak_filtered = strong_weak_pivot[
    (strong_weak_pivot['shots_strong'] + strong_weak_pivot['shots_weak']) >= 10
].copy()
```

```python
strong_weak_filtered
```

```python
player_efficiency = df.groupby(['name', 'preferred_foot']).agg(
    total_shots=('goal', 'count'),
    total_goals=('goal', 'sum'),
    total_xg=('xGoal', 'sum'),
    total_efficiency=('efficiency', 'sum'),
    avg_efficiency=('efficiency', 'mean')
).reset_index()
```

```python
player_eff_clean = player_eff_clean[player_eff_clean['total_shots'] >= 10]
```

## 3. Permutation Test and Bootstrap Confidence Interval

In this section, I test the hypothesis that **left-footed players have higher average finishing efficiency than right-footed players**.

### 3.1 Preparing the Data for Testing
From the player-level efficiency table (`player_eff_clean`):

- Extract the average efficiency for left-footed players  
- Extract the average efficiency for right-footed players  

Compute the observed test statistic:

$$\text{Observed Difference} = \bar{E}_{\text{Left}} - \bar{E}_{\text{Right}}$$$\bar{E}_{\text{Left}}$: The sample mean (average) of the variable $E$ for the "Left" group. $\bar{E}_{\text{Right}}$: The sample mean (average) of the variable $E$ for the "Right" group.

This value represents how much better (or worse) left-footed players performed relative to right-footed players.


### 3.2 Permutation Test  
The permutation test simulates what differences we would observe **if footedness had no effect**.

- **Null hypothesis:** Left- and right-footed players come from the same efficiency distribution.
- **Procedure:**
  - Combine all efficiency values into one array.  
  - Shuffle the values and randomly split into “left” and “right” groups of the same sizes as the real data.
  - Compute the difference in means.
  - Repeat this **10,000 times** to form a null distribution.

**P-value calculation:**

$$p = P(\text{perm diff} \geq \text{observed diff})$$

**Result:**

- **Observed difference:** 0.0033  
- **Permutation p-value:** **0.0650**

**Interpretation:**  
A p-value just above 0.05 suggests **weak evidence** that left-footed players finish better. The result is *suggestive but not conventionally significant*.


### 3.3 Bootstrap Confidence Interval  
Bootstrapping helps understand the sampling variability of the difference in means.

- Draw **10,000 bootstrap samples** from each group (Left and Right) with replacement.
- Compute the mean difference for each bootstrap replicate.
- Take the 2.5th and 97.5th percentiles to form the **95% bootstrap CI**.

**Result:**


$$(\text{95% CI} = (-0.0009,\; 0.0076)$$


**Interpretation:**  
- The interval includes zero → we cannot rule out no difference.  
- But the interval leans slightly positive → consistent with a **small left-foot advantage**.  
- Bootstrapping is necessary because:
  - efficiency values are skewed and not normally distributed,  
  - some players have few shots,  
  - CLT assumptions do not fully hold.


### 3.4 Permutation & Bootstrap Visualizations

Two histograms help visualize uncertainty:

1. **Permutation distribution**  
   - Shows how differences behave under the null.
   - The observed difference is slightly right-tail but not extreme.

2. **Bootstrap distribution**  
   - Shows variability in estimated differences.
   - Helps understand the confidence interval visually.

Both plots reinforce the same conclusion:  
If left-footed players are better finishers, the effect is small.


```python
left_foot  = player_eff_clean[player_eff_clean['preferred_foot']=='Left']['avg_efficiency']
right_foot = player_eff_clean[player_eff_clean['preferred_foot']=='Right']['avg_efficiency']

left_mean  = left_foot.mean()
right_mean = right_foot.mean()

print(f"Left-footed mean efficiency: {left_mean:.4f}")
print(f"Right-footed mean efficiency: {right_mean:.4f}")
print(f"Difference (Left - Right): {left_mean - right_mean:.4f}")
```

```python
left_vals  = left_foot.values
right_vals = right_foot.values

obs_diff = left_vals.mean() - right_vals.mean()
print(f"Observed difference: {obs_diff:.4f}")
```

```python
combined = np.concatenate([left_vals, right_vals])
```

```python
n_perm = 10000
perm_diffs = np.zeros(n_perm)


for i in range(n_perm):
    shuffled = np.random.permutation(combined)   
    sample_group1 = shuffled[:len(left_vals)]
    sample_group2 = shuffled[len(left_vals):]
    perm_diffs[i] = np.mean(sample_group1) - np.mean(sample_group2)
```

```python
p_value = np.mean(perm_diffs >= obs_diff)
print(f"P-value: {p_value:.4f}")
```

```python
n_boot = 10000
boot_diffs = np.zeros(n_boot)

np.random.seed(77)
for i in range(n_boot):
    left_star  = np.random.choice(left_vals,  size=len(left_vals),  replace=True)
    right_star = np.random.choice(right_vals, size=len(right_vals), replace=True)
    boot_diffs[i] = left_star.mean() - right_star.mean()
```

```python
ci_low  = np.percentile(boot_diffs, 2.5)
ci_high = np.percentile(boot_diffs, 97.5)

print(f"95% Bootstrap CI: ({ci_low:.4f}, {ci_high:.4f})")
```

```python
plt.figure(figsize=(10,6))
plt.hist(perm_diffs, bins=40, edgecolor='black')
plt.axvline(obs_diff, color='red', linewidth=2, label=f"Observed = {obs_diff:.4f}")

plt.title("Permutation Test Distribution", fontsize=16)
plt.xlabel("Difference in Mean Efficiency (Left - Right)", fontsize=14)
plt.ylabel("Frequency", fontsize=14)
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(boot_diffs, bins=40, edgecolor='black')
plt.axvline(ci_low,  color='green', linewidth=2, label=f"Lower CI = {ci_low:.4f}")
plt.axvline(ci_high, color='green', linewidth=2, label=f"Upper CI = {ci_high:.4f}")

plt.title("Bootstrap Distribution of Difference in Means", fontsize=16)
plt.xlabel("Difference in Mean Efficiency (Left - Right)", fontsize=14)
plt.ylabel("Frequency", fontsize=14)
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

## 4. Additional Analysis: Strong Foot vs Weak Foot Performance

To further validate the results of the main hypothesis test, I also examined finishing efficiency separately for each player’s **strong foot** and **weak foot**.

This helps answer two questions:

1. **Do players finish better on their strong foot than their weak foot?**  
2. **Do left-footed players outperform right-footed players even when comparing only strong-foot shots?**


### 4.1 Overall Strong vs Weak Foot Efficiency

I first compared the overall averages:

| Foot Type | Mean Efficiency |
|----------|-----------------|
| **Strong Foot** | -0.00108 |
| **Weak Foot**   | -0.01716 |

**Interpretation:**  
Players finish noticeably worse on their weak foot, which is expected and serves as a good validation check of the dataset.


### 4.2 Strong Foot: Left-Footed vs Right-Footed Players

Next, I looked only at *strong-foot* shots and compared left-footed vs right-footed players:

| Group | Mean Strong-Foot Efficiency |
|--------|-----------------------------|
| **Left-Footed Players**  | 0.00021 |
| **Right-Footed Players** | -0.00154 |

**Interpretation:**  
Even when isolating strong-foot shots only, left-footed players still show slightly higher finishing efficiency.


### 4.3 Weak Foot: Left-Footed vs Right-Footed Players

Finally, I compared weak-foot finishing:

| Group | Mean Weak-Foot Efficiency |
|--------|---------------------------|
| **Left-Footed Players**  | -0.01280 |
| **Right-Footed Players** | -0.01871 |

**Interpretation:**  
Weak-foot efficiency is worse for both groups, but right-footed players appear to struggle slightly more on their weaker foot.


### 4.4 Summary of Additional Findings

- Players clearly finish better using their strong foot.  
- **Left-footed players outperform right-footed players** on:
  - overall average efficiency,  
  - strong-foot efficiency,  
  - weak-foot efficiency.  
- These results **support the main hypothesis**, although the effects remain **small**, it is consistent with the permutation test and bootstrap CI.

This complementary analysis strengthens confidence in the conclusion that **left-footed players may have a slight finishing advantage**, though the difference is not large enough to be considered definitively significant.
